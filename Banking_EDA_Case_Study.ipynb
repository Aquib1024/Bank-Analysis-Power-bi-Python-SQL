```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('/content/Banking.csv')
df.head(5)
```

```python
df.shape
```

```python
df.info()
```

```python
# Descriptive Stats of Dataframe
df.describe()
```

```python
# Created bins for income
bins = [0, 100000, 300000, float('inf')]
labels = ['Low', 'Med', 'High']

df['Income Band'] = pd.cut(df['Estimated Income'], bins=bins, labels=labels, right=False)
```

```python
# Plotted Bar Chart
df['Income Band'].value_counts().plot(kind='bar')
```

```python
# Examining the distribution of unique categories in categorical columns

categorical_cols = df[["BRId", "GenderId", "IAId", "Amount of Credit Cards", "Nationality", "Occupation", "Fee Structure", "Loyalty Classification", "Properties Owned", "Risk Weighting", "Income Band"]].columns

for col in categorical_cols:
  print(f"Value Counts for ' {col} ':")
  display(df[col].value_counts())
```

<H1>Univariate Analysis</H1>

```python
# Bar Charts excluding Occupation

other_cols = [col for col in categorical_cols if col != 'Occupation']

plt.figure(figsize=(15, 17))
for i, col in enumerate(other_cols):
    plt.subplot(4, 3, i + 1)
    sns.countplot(data=df, y=col, order=df[col].value_counts().index)
    plt.title(f'Count of {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
plt.tight_layout()
plt.show()

# Occupation Bar chart
plt.figure(figsize=(10, 40))
sns.countplot(data=df, y='Occupation', order=df['Occupation'].value_counts().index)
plt.title('Count of Occupation')
plt.xlabel('Count')
plt.ylabel('Occupation')
plt.yticks(fontsize=8)
plt.tight_layout()
plt.show()
```

<H1>Bivariate Analysis</H1>

```python
for i, predictor in enumerate(df[["BRId", "GenderId", "IAId", "Amount of Credit Cards", "Nationality", "Occupation", "Fee Structure", "Loyalty Classification", "Properties Owned", "Risk Weighting", "Income Band"]].columns):
  plt.figure(i)
  sns.countplot(data=df, x=predictor, hue='GenderId')
```

```python
# Histogram of value counts for different Occupation

for col in categorical_cols:
  if col == "Occupation":
    continue
  plt.figure(figsize=(8,4))
  sns.histplot(df[col])
  plt.title('Histogram of Occupation Count')
  plt.xlabel(col)
  plt.ylabel("Count")
  plt.show()
```

## Numerical Analysis

```python
numerical_cols = ['Estimated Income', 'Superannuation Savings', 'Credit Card Balance', 'Bank Loans', 'Bank Deposits', 'Checking Accounts', 'Saving Accounts', 'Foreign Currency Account', 'Business Lending']

# Univariate analysis and visualization
plt.figure(figsize=(15, 12))
for i, col in enumerate(numerical_cols):
    # Create a 3x3 grid and select the (i+1)-th subplot
    plt.subplot(3, 3, i + 1)
    sns.histplot(df[col], kde=True)
    plt.title(f'Histogram of {col}')

# Add tight_layout to prevent titles/labels from overlapping
plt.tight_layout()
plt.show()
```

# Heatmaps

```python
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(12,12))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

# Overall Data Profile
* **Completeness:** The dataset is perfectly clean, with 3,000 entries and zero missing values across all 25 columns.
* **Client Demographics:** The average client age is 51 years. The `` `GenderId` `` is almost perfectly balanced, with `` `GenderId 2` `` (50.4%) slightly outnumbering `` `GenderId 1` `` (49.6%). {GenderId 2 is Female and GenderId 1 is male}
* **Income Distribution:** The client base is primarily in the 'Medium' income bracket (\$100k - \$300k), accounting for 50.6% of clients. 'Low' income (<\$100k) clients are the next largest group (34.2%), while 'High' income (>\$300k) clients are the minority (15.2%).
* **Geographic/Nationality:** The client base is diverse, but heavily weighted towards 'European' (43.6%), followed by 'Asian' (25.1%) and 'American' (16.9%).
* **Occupations:** This is a **high-cardinality** field with 195 unique occupations. The distribution is extremely fragmented; the most common job ("Structural Analysis Engineer") represents less than 1% of the client base.

---

# Categorical Insights (Univariate)
* **Product Holding:** Most clients (64.1%) hold only **one credit card**. 25.5% hold two, and only 10.4% hold three.
* **Client Loyalty:** 'Jade' (44.4%) is by far the most common loyalty tier, followed by 'Silver' (25.6%). 'Platinum' (10.6%) is the least common.
* **Property Ownership:** The number of properties owned is very evenly distributed, with 23-26% of clients in each bucket (0, 1, 2, or 3 properties).
* **Risk Profile:** The majority of clients are classified as 'Risk Weighting 2' (40.7%) or 'Risk Weighting 1' (27.9%), suggesting a generally low-to-moderate risk profile for the portfolio.

---

# Bivariate Insights (Analysis by Gender)
* **No Major Gender Disparity:** For most high-level metrics (`` `Income Band` ``, `` `Fee Structure` ``, `` `Risk Weighting` ``, `` `Loyalty Classification` ``), there is **no significant skew by `` `GenderId` ``**. All categories are within a 47-53% split, indicating that gender is not a strong predictor for these top-level classifications.
* **Occupation and Gender:** This is the **only** area with strong gender-based trends. The analysis shows some occupations are highly skewed:
    * **Predominantly `` `GenderId 2` ``:** 'Nurse' (91.7%), 'Design Engineer' (82.4%), 'Human Resources Assistant I' (68.2%).
    * **Predominantly `` `GenderId 1` ``:** 'Database Administrator IV' (77.8%), 'Nuclear Power Engineer' (76.5%), 'Chief Design Engineer' (72.7%).
* **Minor Skews:**
    * Clients with **2 properties** are slightly more likely to be `` `GenderId 1` `` (52.4%).
    * Clients in the **'Platinum'** loyalty tier are slightly more likely to be `` `GenderId 2` `` (53.3%).

---

# Numerical Insights (Correlation Heatmap)
* **Strongest Relationship:** There is a **very strong positive correlation** between `` `Bank Deposits` `` and `` `Checking Accounts` `` (0.84). This is followed by a strong correlation between `` `Bank Deposits` `` and `` `Saving Accounts` `` (0.75). This implies that clients do not just let deposits sit; they actively move them into their checking and savings accounts.
* **Income Is a Weak Driver:** `` `Estimated Income` `` has only a weak-to-moderate positive correlation (0.26 - 0.37) with all other financial variables. Its strongest relationship is with `` `Superannuation Savings` `` (0.37), which is logical. This suggests that income alone is not the primary driver of a client's loan amounts or deposit balances.
* **Business and Bank Loans:** `` `Business Lending` `` and `` `Bank Loans` `` have a moderate positive relationship (0.42). This suggests clients who take on personal bank loans are also more likely to be engaged in business lending.
* **No Negative Correlations:** All financial metrics are positively correlated. There are no "trade-off" relationships (e.g., where high bank loans would mean low savings). Instead, financial activity begets more financial activity—clients with higher balances or loans in one area tend to have higher balances or loans in others.
